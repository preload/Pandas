```python
import pandas as pd

df_employee = pd.DataFrame({
    'id': [1, 2, 3, 4, 5, 6],
    'name': ['Joe', 'Jim', 'Henry', 'Sam', 'Max', 'Janice'],
    'salary': [70000, 90000, 80000, 60000, 90000, 60000],
    'departmentId': [1, 1, 2, 2, 1, 3]
})

df_department = pd.DataFrame({
    'id': [1, 2, 3],
    'name': ['IT', 'Sales', 'HR']
})
```

```python
def department_highest_salary(employee: pd.DataFrame, department: pd.DataFrame) -> pd.DataFrame:
    df_department.rename(columns={'id':'departmentId', 'name':'Department'}, inplace=True)
    df_employee.rename(columns={'name':'Employee', 'salary':'Salary'}, inplace=True)
    df = df_employee.merge(how='left', left_on='departmentId', right_on='departmentId', right=df_department)
    df.drop(columns=['departmentId','id'], inplace=True)
    df = df.loc[:,['Department', 'Employee', 'Salary']]
    max = df.groupby('Department')['Salary'].transform('max')
    return df.loc[df['Salary'] == max]
```
