```python
import pandas as pd

data = {
    'id': [101, 102, 103, 104, 105, 106],
    'name': ['John', 'Dan', 'James', 'Amy', 'Anne', 'Ron'],
    'department': ['A', 'A', 'A', 'A', 'A', 'B'],
    'managerId': [None, 101.0, 101.0, 101.0, 101.0, 101.0]
}

df = pd.DataFrame(data)
```

```python
def find_managers(df: pd.DataFrame) -> pd.DataFrame:
    if len(df) == 0: return df['name'].to_frame()
    ids = df.managerId.value_counts()
    ids = ids[ids >= 5].index.tolist()
    return df.loc[df.id.isin(ids)]['name'].to_frame()
```

```python
find_managers(df)
```
