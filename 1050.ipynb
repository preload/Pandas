```python
import pandas as pd

df = pd.DataFrame({
    'actor_id': [1, 1, 1, 1, 1, 2, 2],
    'director_id': [1, 1, 1, 2, 2, 1, 1],
    'timestamp': [0, 1, 2, 3, 4, 5, 6]
})
```

```python
def actors_and_directors(df: pd.DataFrame) -> pd.DataFrame:
    return df.groupby(['actor_id','director_id'],as_index=False).count().loc[lambda x: x['timestamp'] >= 3].set_index('timestamp')
```

```python
actors_and_directors(df)
```
