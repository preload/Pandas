```python
import pandas as pd

def big_countries(df: pd.DataFrame) -> pd.DataFrame:
    return df.loc[((df.population > 25000000) | (df.area > 3000000)), ['name','population','area']]
```

```python
import pandas as pd

def find_products(df: pd.DataFrame) -> pd.DataFrame:
    return df.loc[((df.low_fats == 'Y') & (df.recyclable == 'Y')) , ['product_id']]
```

```python
import pandas as pd

def find_customers(df1: pd.DataFrame, df2: pd.DataFrame) -> pd.DataFrame:
    return df1.loc[(df1.any(df2, axis=1)),['name']]
```
