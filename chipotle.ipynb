```python
df.item_price =  df.item_price.apply(lambda x: float(x.replace('$','')))
df['single_item_price'] = df.item_price / df.quantity
```

```python
df.loc[df['single_item_price'] > 10].drop_duplicates(subset=['item_name','choice_description'], keep='first')
```
