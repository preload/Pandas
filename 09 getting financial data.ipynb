```python
import pandas as pd
from datetime import datetime
import yfinance as yf
```

```python
df_apple = yf.download('AAPL', pd.to_datetime('01/01/2015'), pd.to_datetime(datetime.today()))
df_apple['stock'] = 'AAPL'
df_apple = df_apple.set_axis(['Close', 'High', 'Low', 'Open', 'Volume','stock'], axis=1)
```

```python
df_tesla = yf.download('TSLA', pd.to_datetime('01/01/2015'), pd.to_datetime(datetime.today()))
df_tesla['stock'] = 'TSLA'
df_tesla = df_tesla.set_axis(['Close', 'High', 'Low', 'Open', 'Volume','stock'], axis=1)
```

```python
df_ibm = yf.download('IBM', pd.to_datetime('01/01/2015'), pd.to_datetime(datetime.today()))
df_ibm['stock'] = 'IBM'
df_ibm = df_ibm.set_axis(['Close', 'High', 'Low', 'Open', 'Volume','stock'], axis=1)
```

```python
df_mfst = yf.download('MSFT', pd.to_datetime('01/01/2015'), pd.to_datetime(datetime.today()))
df_mfst['stock'] = 'MFST'
df_mfst = df_mfst.set_axis(['Close', 'High', 'Low', 'Open', 'Volume','stock'], axis=1)
```

```python
#step 7
df = pd.concat([df_apple, df_tesla, df_ibm, df_mfst]).sort_values('Date')
df
```

```python
#step 8
df.set_index([df.index,'stock'], inplace=True)
df
```

```python
#step 9
vol = df['Volume']
```

```python
#step 10
vol.unstack().groupby(pd.Grouper(freq='W')).sum()
```

```python
#step 10 v2
vol.unstack().resample('W').sum()
```

```python
df_vol = vol.unstack()

df_vol.groupby([
    df_vol.index.year.rename('year'),
    df_vol.index.isocalendar().week.rename('week')
]).sum()
```
