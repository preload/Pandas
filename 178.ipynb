```python
import pandas as pd

df = pd.DataFrame({
    'id': [1, 2, 3, 4, 5, 6],
    'score': [3.50, 3.65, 4.00, 3.85, 4.00, 3.65]
})

df2 = pd.DataFrame()
```

```python
def order_scores(df: pd.DataFrame) -> pd.DataFrame:
    if len(df) == 0 : return pd.DataFrame(columns=['score','rank'])
    df['rank'] = df['score'].rank(method='dense', ascending=False).astype(int)
    return df[['score','rank']].sort_values('score', ascending=False)
```

```python
order_scores(df)
```
