```python
import pandas as pd

def valid_emails(df: pd.DataFrame) -> pd.DataFrame:
    pattern = re.compile(r'^[a-zA-Z]+[\w\.\-]*@leetcode\.com$')
    valid = df.mail.str.match(pattern)
    return df[valid]
```
