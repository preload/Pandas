```python
import pandas as pd

df = pd.DataFrame({
    'date_id': pd.to_datetime(['2020-12-8', '2020-12-8', '2020-12-8', '2020-12-7', '2020-12-7', '2020-12-8', '2020-12-8', '2020-12-7', '2020-12-7', '2020-12-7']),
    'make_name': ['toyota', 'toyota', 'toyota', 'toyota', 'toyota', 'honda', 'honda', 'honda', 'honda', 'honda'],
    'lead_id': [0, 1, 1, 0, 0, 1, 2, 0, 1, 2],
    'partner_id': [1, 0, 2, 2, 1, 2, 1, 1, 2, 1]
})
```

```python
def daily_leads_and_partners(df: pd.DataFrame) -> pd.DataFrame:
    return df.groupby(['date_id','make_name'], as_index=False).nunique().rename(columns={'lead_id':'unique_leads', 'partner_id':'unique_partners'})
```

```python
daily_leads_and_partners(df)
```
