```python
import pandas as pd

df1 = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'name': ['Joe', 'Henry', 'Sam', 'Max']
})

df2 = pd.DataFrame({
    'id': [1, 2],
    'customerId': [3, 1]
})

def find_customers(df1: pd.DataFrame, df2: pd.DataFrame) -> pd.DataFrame:
    no_orders = [int(value) for value in df1.id.values if value not in df2.customerId.values]
    return df1.loc[df1.id.isin(no_orders)]['name'].to_frame().rename(columns={'name': 'Customers'})
```

```python
find_customers(df1, df2)
```
