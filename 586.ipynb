```python
import pandas as pd

df = pd.DataFrame({
    'order_number': [1, 2, 3, 4],
    'customer_number': [1, 2, 3, 3]
})
```

```python
def largest_orders(df: pd.DataFrame) -> pd.DataFrame:
    return df.groupby('customer_number', as_index=False).count().loc[lambda x: x['order_number'] == x.order_number.max()]['customer_number']
```

```python
largest_orders(df)
```
