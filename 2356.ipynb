```python
import pandas as pd

df = pd.DataFrame({
    'teacher_id': [1, 1, 1, 2, 2, 2, 2],
    'subject_id': [2, 2, 3, 1, 2, 3, 4],
    'dept_id': [3, 4, 3, 1, 1, 1, 1]
})
```

```python
def count_unique_subjects(df: pd.DataFrame) -> pd.DataFrame:
    return df.groupby('teacher_id')['subject_id'].nunique().to_frame().reset_index().rename(columns={'subject_id': 'cnt'})
```

```python
count_unique_subjects(df)
```
