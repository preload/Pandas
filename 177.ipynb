```python
import pandas as pd
from pandas.conftest import ascending

df = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'salary': [100, 200, 300, 300]
})
```

```python
def nth_highest_salary(df: pd.DataFrame, N: int) -> pd.DataFrame:
    sals = sorted(list(set(df.salary)), reverse=True)
    val = pd.NA if len(sals) < N or N < 1 else sals[N-1]
    return pd.DataFrame({f'getNthHighestSalary({N})': [val]})
```

```python
nth_highest_salary(df,-1)
```
