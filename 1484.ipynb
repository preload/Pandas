```python
import pandas as pd

df = pd.DataFrame({
    'sell_date': pd.to_datetime(['2020-05-30', '2020-06-01', '2020-06-02', '2020-05-30', '2020-06-01', '2020-06-02', '2020-05-30']),
    'product': ['Headphone', 'Pencil', 'Mask', 'Basketball', 'Bible', 'Mask', 'T-Shirt']
})
```

```python
def categorize_products(df: pd.DataFrame) -> pd.DataFrame:
    df['num_sold'] = 1
    df = df.sort_values(by='product')
    df['product'] = ',' + df['product']
    # df.loc[df.duplicated(keep='last'), 'num_sold'] += 1
    df.drop_duplicates(subset=['product', 'sell_date'], keep='first', inplace=True)
    df = df.groupby('sell_date', as_index=False).sum()
    df['product'] = df['product'].str.removeprefix(',')
    return df.loc[:, ['sell_date','num_sold', 'product']].rename(columns={'product':'products'})
```

```python
categorize_products(df)
```
