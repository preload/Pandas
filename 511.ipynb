```python
import pandas as pd

df = pd.DataFrame({
    'player_id': [1, 1, 2, 3, 3],
    'device_id': [2, 2, 3, 1, 4],
    'event_date': pd.to_datetime(['2016-03-01', '2016-05-02', '2017-06-25', '2016-03-02', '2018-07-03']),
    'games_played': [5, 6, 1, 0, 5]
})
```

```python
def game_analysis(df: pd.DataFrame) -> pd.DataFrame:
    return df.groupby('player_id')['event_date'].min().reset_index().rename(columns={'event_date': 'first_login'})
```

```python
game_analysis(df)
```
