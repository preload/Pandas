```python
import pandas as pd

df = pd.DataFrame({
    'account_id': [3, 2, 8, 6],
    'income': [108939, 12747, 87709, 91796]
})
```

```python
def count_salary_categories(df: pd.DataFrame) -> pd.DataFrame:
    low_salary = 0
    medium_salary = 0
    high_salary = 0

    for income in df['income']:
        if income < 20000:
            low_salary += 1
        elif income <= 50000:
            medium_salary += 1
        else:
            high_salary += 1

    return pd.DataFrame(
        {
            'category': ['Low Salary', 'Average Salary', 'High Salary'],
            'accounts_count': [low_salary, medium_salary, high_salary]
        }
    ).sort_values('accounts_count', ascending=False)
```

```python
count_salary_categories(df)
```
