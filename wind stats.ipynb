```python
import pandas as pd
from jupyter_core.migrate import regex
from numpy.ma.core import std
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/06_Stats/Wind_Stats/wind.data', delimiter=r'\s+')
```

```python
df
```

```python
#step 4
df['Yr'] = df['Yr']+1900
```

```python
#step 3
df['Yr'] = df['Yr'].astype(str)
df['Mo'] = df['Mo'].astype(str)
df['Dy'] = df['Dy'].astype(str)
df['Yr'] = df['Yr'] + '-' + df['Mo'] + '-' + df['Dy']
del df['Date']
del df['Mo']
del df['Dy']
df.rename({'Yr':'Date'}, axis=1, inplace=True)
```

```python
#step 4
df['Date'] = pd.to_datetime(df['Date'], format='%Y-%m-%d')
```

```python
#step 5
df = df.set_index('Date')
```

```python
df
```

```python
#step 6
df.isna().sum()
```

```python
#step 7
df.isna().sum().sum()
```

```python
#step 8
df.mean().mean()
```

```python
#step 9
df.agg(['min', 'max', 'mean', 'std'])
```

```python
#step 10
df.agg(['min', 'max', 'mean', 'std'], axis=1)
```

```python
#step 11
only_jan = df.loc[df.index.month == 1]
only_jan.mean()
```

```python
#step 12
df.groupby(df.index.year).mean()
```

```python
#step 13
df.groupby(df.index.to_period('M')).mean()
```

```python
#step 14
df.groupby(df.index.to_period('W')).mean()
```

```python
#step 15
filtered = df.loc[df.index > pd.to_datetime('1961-01-01')]
filtered.groupby(filtered.index.to_period('W')).agg(['min','max','mean','std']).stack().head(52)
```
