```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/04_Apply/US_Crime_Rates/US_Crime_Rates_1960_2014.csv')
df.head(2)
```

```python
#step 4
df.dtypes
```

```python
#step 5
df['Year'] = pd.to_datetime(df['Year'], format='%Y')
```

```python
#step 6
df.set_index('Year', inplace=True)
```

```python
#step 7
df.drop('Total', axis=1, inplace=True)
```

```python
agg_dict = { col: ('sum' if col != 'Population' else 'max') for col in df.columns }

df.groupby((df.index.year //10 * 10)).agg(agg_dict)
```
