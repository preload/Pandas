```python
import pandas as pd

df = pd.DataFrame({
    'employee_id': [2, 3, 7, 8, 9],
    'name': ['Meir', 'Michael', 'Addilyn', 'Juan', 'Kannon'],
    'salary': [3000, 3800, 7400, 6100, 7700]
})

def calculate_special_bonus(df: pd.DataFrame) -> pd.DataFrame:
    get_bonus =  df.loc[(df.employee_id % 2 != 0) & (df.name.str[0] != 'M')]
    df['bonus'] = 0
    df.loc[get_bonus.index, 'bonus'] = df.loc[get_bonus.index, 'salary']
    return df.loc[:,['employee_id', 'bonus']].sort_values(by='employee_id', ascending=True)
```

```python
calculate_special_bonus(df)
```
