```python
import pandas as pd

df = pd.DataFrame({
    'emp_id': [1, 1, 1, 2, 2],
    'event_day': ['2020-11-28', '2020-11-28', '2020-12-03', '2020-11-28', '2020-12-09'],
    'in_time': [4, 55, 1, 3, 47],
    'out_time': [32, 200, 42, 33, 74]
})
```

```python
def total_time(df: pd.DataFrame) -> pd.DataFrame:
    df['total_time'] = df['out_time'] - df['in_time']
    return df.groupby(['event_day','emp_id'], as_index=False).agg(total_time = ('total_time', 'sum')).rename(columns={'event_day':'day'})
```

```python
def total_time(df: pd.DataFrame) -> pd.DataFrame:
    return (df['out_time'] - df['in_time']).groupby([df['event_day'],df['emp_id']]).sum().reset_index(name='total_time').rename(columns={'event_day':'day'})
```
