```python
import pandas as pd

df = pd.DataFrame({
    'id': [1, 2, 3],
    'email': ['john@example.com', 'bob@example.com', 'john@example.com']
})

def delete_duplicate_emails(dfs):
    dfs = dfs.sort_values(by='id')
    dfs = dfs.drop_duplicates(subset=['email'], keep='first')
```

```python
delete_duplicate_emails(df)
df
```
