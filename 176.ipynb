```python
import pandas as pd

df = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'salary': [100, 200, 300, 300]
})


def second_highest_salary(df: pd.DataFrame) -> pd.DataFrame:
    sals = sorted(list(set(df.salary)), reverse=True)
    val = pd.NA if len(sals) < 2 else sals[1]
    return pd.DataFrame({f'SecondHighestSalary': [val]})
```

```python
second_highest_salary(df)
```
