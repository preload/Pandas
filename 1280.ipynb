```python
import pandas as pd

students = pd.DataFrame({
    'student_id': [1, 2, 13, 6],
    'student_name': ['Alice', 'Bob', 'John', 'Alex']
})

subjects = pd.DataFrame({
    'subject_name': ['Math', 'Physics', 'Programming']
})

examinations = pd.DataFrame({
    'student_id': [1, 1, 1, 2, 1, 1, 13, 13, 13, 2, 1],
    'subject_name': ['Math', 'Physics', 'Programming', 'Programming', 'Physics', 'Math', 'Math', 'Programming', 'Physics', 'Math', 'Math']
})
```

```python
df = students.merge(subjects, how='cross')
df['count'] = 0
examinations = examinations.groupby(['student_id', 'subject_name'], as_index=False).value_counts()
df = df.merge(examinations, how='left', on=['student_id', 'subject_name', 'student_id'])
del df['count_x']
df.rename(columns={'count_y': 'attended_exams'}, inplace=True)
df['attended_exams'].fillna(0, inplace=True)
df['attended_exams'] = df['attended_exams'].astype(int)
df.sort_values(['student_id', 'subject_name'])
```
