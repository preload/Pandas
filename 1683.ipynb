```python
import pandas as pd

df = pd.DataFrame({
    'tweet_id': [1, 2],
    'content': ['Let us Code', 'More than fifteen chars are here!']
})

def invalid_tweets(df: pd.DataFrame) -> pd.DataFrame:
    return df.loc[df.content.str.len() > 15, 'tweet_id'].to_frame()
```

```python
invalid_tweets(df)
```
