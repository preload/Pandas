```python
import pandas as pd

df = pd.DataFrame({
    'student': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I'],
    'class': ['Math', 'English', 'Math', 'Biology', 'Math', 'Computer', 'Math', 'Math', 'Math']
})
```

```python
def find_classes(df: pd.DataFrame) -> pd.DataFrame:
    pre_output = df.groupby('class', as_index=False).nunique()
    return pre_output.loc[pre_output['student'] >= 4, 'class']
```

```python
def find_classes(df: pd.DataFrame) -> pd.DataFrame:
    return df.groupby('class', as_index=False).nunique().loc[lambda x: x['student'] >= 4, 'class']
```

```python
find_classes(df)
```
