```python
import pandas as pd

df = pd.DataFrame({
    'product_id': [0, 1],
    'store1': [95, 70],
    'store2': [100, None],
    'store3': [105, 80]
})


def rearrange_products_table(df: pd.DataFrame) -> pd.DataFrame:
    df =  df.melt(id_vars=['product_id'], value_vars=['store1', 'store2', 'store3'], var_name='store', value_name='price')
    df.dropna(inplace=True)
    return df
```

```python
rearrange_products_table(df)
```
