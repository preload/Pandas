```python
import pandas as pd

sales_person_data = {
    'sales_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Amy', 'Mark', 'Pam', 'Alex'],
    'salary': [100000, 12000, 65000, 25000, 5000],
    'commission_rate': [6, 5, 12, 25, 10],
    'hire_date': pd.to_datetime(['2006-04-01', '2010-05-01', '2008-12-25', '2005-01-01', '2007-02-03'])
}
sales_person_df = pd.DataFrame(sales_person_data)

company_data = {
    'com_id': [1, 2, 3, 4],
    'name': ['RED', 'ORANGE', 'YELLOW', 'GREEN'],
    'city': ['Boston', 'New York', 'Boston', 'Austin']
}
company_df = pd.DataFrame(company_data)

orders_data = {
    'order_id': [1, 2, 3, 4],
    'order_date': pd.to_datetime(['2014-01-01', '2014-02-01', '2014-03-01', '2014-04-01']),
    'com_id': [3, 4, 1, 1],
    'sales_id': [4, 5, 1, 4],
    'amount': [10000, 5000, 50000, 25000]
}
orders_df = pd.DataFrame(orders_data)
```

```python
def sales_person(sales_person: pd.DataFrame, company: pd.DataFrame, orders: pd.DataFrame) -> pd.DataFrame:
    if company_df['name'].str.match(r'^RED$').sum() != 1: return sales_person['name'].to_frame()
    company_id = company.loc[company['name'] == 'RED', 'com_id'].values[0]
    sales_ids = orders.loc[orders.com_id == company_id]['sales_id'].values
    sales_person_df = sales_person.loc[~sales_person['sales_id'].isin(sales_ids)]
    return sales_person_df['name'].to_frame()
```

```python
sales_person(sales_person_df, company_df, orders_df)
```
