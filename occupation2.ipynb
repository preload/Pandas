```python
import pandas as pd
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user', delimiter='|')
```

```python
df.head()
```

```python
df.groupby('occupation')['age'].mean()
```

```python
(df.groupby('occupation')['gender']
 .value_counts(normalize=True, sort=False)
 .loc[:,'M']
 .sort_values(ascending=False)
 ).apply(lambda x : x*100)
```

```python
df.groupby('occupation')['age'].agg([min,max])
```

```python
df.groupby(['occupation','gender'])['age'].mean()
```

```python
(
    df.groupby('occupation')['gender']
    .value_counts(normalize=True, sort=False)
    .to_frame('percent')
).apply(
    lambda x: round(x*100).astype(int)
)
```
