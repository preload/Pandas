```python
import pandas as pd
import re

df = pd.DataFrame({
    'patient_id': [1, 2, 3, 4, 5],
    'patient_name': ['Daniel', 'Alice', 'Bob', 'George', 'Alain'],
    'conditions': ['SADIAB100', '', 'DIAB100 MYOP', 'ACNE DIAB100', 'ACNE +DIAB100']
})
```

```python
def check_condition(x):
    for xx in x.split():
        if xx.startswith('DIAB1'): return True
    return False

def find_patients(df: pd.DataFrame) -> pd.DataFrame:
    return df.loc[df.conditions.map(check_condition)]
```

```python
find_patients(df)
```
